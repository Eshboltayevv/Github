26. Yangilangan datafreym uchun region bo'yicha to'ldirilmagan qiymatlar sonini chiqaring.

```python
# javobni shu yerga yozing...
import pandas as pd

df = pd.read_csv('cars_data_filtered.csv')


nan_by_region = df.groupby('region', group_keys=False).apply(
    lambda x: x.isnull().sum(),
    include_groups=False
)

print("Region bo'yicha to'ldirilmagan qiymatlar soni:")
print("=" * 60)
print(nan_by_region)

nan_total_by_region = df.groupby('region', group_keys=False).apply(
    lambda x: x.isnull().sum().sum(),
    include_groups=False
)

print("\n" + "=" * 60)
print("Har bir region uchun JAMI to'ldirilmagan qiymatlar soni:")
print("=" * 60)
print(nan_total_by_region)

```

27. Datafreym haqida ma'lumotni chiqaring va ma'lumot turlariga e'tibor bering.

```python
# javobni shu yerga yozing...
import pandas as pd

df = pd.read_csv('cars_data_filtered.csv')

print("=" * 60)
print("1. UMUMIY MA'LUMOT")
print("=" * 60)
print(f"📊 Qatorlar soni: {df.shape[0]}")
print(f"📋 Ustunlar soni: {df.shape[1]}")

print("\n" + "=" * 60)
print("2. MA'LUMOT TURLARI (dtypes)")
print("=" * 60)
print(df.dtypes)

print("\n" + "=" * 60)
print("3. BATAFSIL MA'LUMOT (info)")
print("=" * 60)
df.info()

print("\n" + "=" * 60)
print("4. STATISTIK TAVSIF (describe)")
print("=" * 60)
print(df.describe())

print("\n" + "=" * 60)
print("5. MA'LUMOT TURLARI BO'YICHA GURUHLASH")
print("=" * 60)

for dtype in df.dtypes.unique():
    cols = df.select_dtypes(include=[dtype]).columns.tolist()
    print(f"\n🔹 {dtype} ({len(cols)} ta ustun):")
    for col in cols:
        print(f"   • {col}")

print("\n" + "=" * 60)
print("6. XOTIRA ISHLATILISHI")
print("=" * 60)
print(f"Xotira hajmi: {df.memory_usage(deep=True).sum() / 1024:.2f} KB")

print("\n" + "=" * 60)
print("7. TO'LDIRILMAGAN QIYMATLAR")
print("=" * 60)
print(df.isnull().sum())
```


31. Interkartil oralig'idan foydalanib, normal narx qiymatlari uchun chegaralarni belgilang.


```python
# javobni shu yerga yozing...
import pandas as pd

df = pd.read_csv('cars_data.csv')

Q1 = df['price'].quantile(0.25)
Q3 = df['price'].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print(f"Q1 (25%): {Q1}")
print(f"Q3 (75%): {Q3}")
print(f"IQR: {IQR}")
print(f"\nNormal narx chegaralari:")
print(f"Pastki chegara: {lower_bound}")
print(f"Yuqori chegara: {upper_bound}")
```


32.  Avtomobil narxidagi chiqindilar miqdorini hisoblang va chiqarib oling .


```python
# javobni shu yerga yozing...
import pandas as pd

df = pd.read_csv('cars_data.csv')

Q1 = df['price'].quantile(0.25)
Q3 = df['price'].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[(df['price'] < lower_bound) | (df['price'] > upper_bound)]

print(f"Chiqindilar miqdori: {len(outliers)}")
print(f"Jami qatorlar: {len(df)}")
print(f"Chiqindilar foizi: {len(outliers)/len(df)*100:.2f}%")

print("\nChiqindilar:")
print(outliers[['price']])
```

33. Avtomobil narxidagi chiqindilarning ulushini hisoblang va oling

```python
# javobni shu yerga yozing...
import pandas as pd

df = pd.read_csv('cars_data.csv')

Q1 = df['price'].quantile(0.25)
Q3 = df['price'].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[(df['price'] < lower_bound) | (df['price'] > upper_bound)]

outlier_percentage = (len(outliers) / len(df)) * 100

print(f"Chiqindilar soni: {len(outliers)}")
print(f"Jami qatorlar: {len(df)}")
print(f"Chiqindilarning ulushi: {outlier_percentage:.2f}%")
```

34.  Chiqindilarni va bosh qiymatli qatorlarni olib tashlang va saqlang.

```python
# javobni shu yerga yozing...
import pandas as pd

df = pd.read_csv('cars_data.csv')

df_clean = df.dropna()

Q1 = df_clean['price'].quantile(0.25)
Q3 = df_clean['price'].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR


df_filtered = df_clean[(df_clean['price'] >= lower_bound) & (df_clean['price'] <= upper_bound)]

df_filtered.to_csv('cars_data_filtered.csv', index=False)


print(f"Asl qatorlar: {len(df)}")
print(f"Bosh qiymatlar olib tashlangandan so'ng: {len(df_clean)}")
print(f"Chiqindilar olib tashlangandan so'ng: {len(df_filtered)}")
print(f"Saqlangan fayl: cars_data_filtered.csv")
```
