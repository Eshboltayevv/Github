### 2. Kategoriyali o‘zgaruvchilarni qayta ishlash
- `model`, `manufacturer`, `fuel_type`, `transmission`, `region` kategoriyali o‘zgaruvchilarni raqamli belgilariga aylantirish uchun `OneHotEncoder` dan foydalaning.
- Hosil qilingan raqamli belgilarni asosiy datafreymga qo‘shing.


```python
import pandas as pd
from sklearn. preprocessing import OneHotEncoder


df = pd.DataFrame({
    "model": ["Camry", "Corolla", "BMW X5"],
    "manufacturer": ["Toyota", "Toyota", "BMW"],
    "fuel_type": ["Petrol", "Hybrid", "Diesel"],
    "transmission": ["Automatic", "Automatic", "Manual"],
    "region": ["Asia", "Asia", "Europe"]
})


categorical_cols = ["model", "manufacturer", "fuel_type", "transmission", "region"]


encoder = OneHotEncoder(sparse_output=False)


encoded_array = encoder.fit_transform(df[categorical_cols])


encoded_df = pd.DataFrame(
    encoded_array,
    columns=encoder.get_feature_names_out(categorical_cols)
)


df_encoded = pd.concat([df.drop(columns=categorical_cols), encoded_df], axis=1)


print(df_encoded)
```

### 3. Ma'lumotlarni standartlashtirish
- `price`, `engine_power`, `odometr` kabi raqamli o‘zgaruvchilarni `StandardScaler` yordamida standartlashtiring.
- Har bir o‘zgaruvchi uchun standartlashtirilgan qiymatlarni yangi ustunga qo‘shing.


```python
import pandas as pd
from sklearn. preprocessing import StandardScaler


df = pd.DataFrame({
    "price": [10000, 15000, 20000],
    "engine_power": [90, 110, 150],
    "odometr": [50000, 30000, 10000]
})


num_cols = ["price", "engine_power", "odometr"]


scaler = StandardScaler()


scaled_values = scaler.fit_transform(df[num_cols])


df_scaled = df.copy()

df_scaled["price_scaled"] = scaled_values[:, 0]
df_scaled["engine_power_scaled"] = scaled_values[:, 1]
df_scaled["odometr_scaled"] = scaled_values[:, 2]


print(df_scaled)
```

### 4. MinMax normalizatsiyasi
- `price`, `engine_power`, `odometer` o‘zgaruvchilarini `MinMaxScaler` yordamida normalizatsiyadan o‘tkazing. Normalizatsiyalangan qiymatlarni yangi ustunlarga saqlang.


```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler


df = pd.DataFrame({
    "price": [10000, 15000, 20000],
    "engine_power": [90, 110, 150],
    "odometer": [50000, 30000, 10000]
})


num_cols = ["price", "engine_power", "odometer"]


scaler = MinMaxScaler()


scaled_values = scaler.fit_transform(df[num_cols])


df_normalized = df.copy()


df_normalized["price_norm"] = scaled_values[:, 0]
df_normalized["engine_power_norm"] = scaled_values[:, 1]
df_normalized["odometer_norm"] = scaled_values[:, 2]


print(df_normalized)
```

### 5. Sanalarga asoslangan belgilar
- `year` o‘zgaruvchisidan avtomobilning ishlab chiqarilgan yilini oling va avtomobilning hozirgi yilga nisbatan yoshini o‘z ichiga oladigan yangi belgi yarating.


```python
import pandas as pd
from datetime import datetime


df = pd.DataFrame({
    "year": [2010, 2015, 2020, 2023]
})


current_year = datetime.now().year

df["car_age"] = current_year - df["year"]


print(df)
```
