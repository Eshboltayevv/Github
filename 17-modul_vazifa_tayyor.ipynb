7. Narx kategoriyalarining ma'lumotlar to'plamidagi ulushlarini chiqaring.



```python
# javobingiz bu yerda...

bins = [0, 10000, 20000, df['price'].max()]

labels = ['arzon', 'o‘rta', 'qimmat']

df['price_category'] = pd.cut(df['price'], bins=bins, labels=labels)

print(df['price_category'].value_counts(normalize=True) * 100)

```

8. Maqsadli o'zgaruvchining qiymatlari qanday taqsimlanganligini tahlil qiling (masalan, taqsimot tengmi).

```python
# javobingiz bu yerda...
y = df['price']
import matplotlib.pyplot as plt

plt.hist(y, bins=20, edgecolor='black')
plt.title("Maqsadli o'zgaruvchi (Price) taqsimoti")
plt.xlabel("Price")
plt.ylabel("Qiymatlar soni")
plt.show()
print("O'rtacha:", y.mean())
print("Median:", y.median())
print("Standart og'ish:", y.std())
print("Minimum:", y.min())
print("Maximum:", y.max())
```
