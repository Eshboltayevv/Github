 2. Narx va dvigatel hajmi o'rtasidagi bog'liqlik:
Avtomobillarning narxi va dvigatel hajmi o'rtasidagi bog'liqlikni tahlil qiling. Scatter plot diagrammasini tuzib, dvigatel hajmi va narxni turli ranglarda ko'rsating.

```python
# Kodni shu yerda yozing
import pandas as pd
import matplotlib.pyplot as plt

# 1. Datasetni yuklash
df = pd.read_csv("cars_data_cleaned.csv")

# 2. Scatter plot (engine_power vs price)
plt.figure()

scatter = plt.scatter(
    df['engine_power'],   # dvigatel quvvati
    df['price'],          # narx
    c=df['price'],        # ranglar narx bo‘yicha
    cmap='viridis',
    alpha=0.7
)

# 3. Colorbar qo‘shish
plt.colorbar(scatter, label="Narx")

# 4. Grafik nomlari
plt.title("Dvigatel quvvati va narx o‘rtasidagi bog‘liqlik")
plt.xlabel("Dvigatel quvvati (engine_power)")
plt.ylabel("Narx")

plt.grid()
plt.show()

# 5. Korrelyatsiya
corr = df[['engine_power', 'price']].corr().iloc[0,1]
print(f"\nKorrelyatsiya koeffitsienti: {corr:.2f}")
```

3. Uzatish qutisi va avtomobil narxi:
Avtomobillarning uzatish qutisi turi (avtomat yoki mexanika) va narxi o'rtasidagi bog'liqlikni o'rganing. Har bir uzatish qutisi turi uchun avtomobillar narxini vizualizatsiya qiling.

```python
# Kodni shu yerda yozing
import pandas as pd
import matplotlib.pyplot as plt

# 1. Datasetni yuklash
df = pd.read_csv("cars_data_cleaned.csv")

# 2. Transmission ustunini kichik harfga o‘tkazamiz (xatolik oldini olish uchun)
df['transmission'] = df['transmission'].str.lower()

# 3. Boxplot (eng professional vizualizatsiya)
plt.figure()

df.boxplot(column='price', by='transmission')

plt.title("Uzatish qutisi va narx o‘rtasidagi bog‘liqlik")
plt.suptitle("")  # default title ni olib tashlaydi
plt.xlabel("Transmission (avtomat / mexanika)")
plt.ylabel("Narx")

plt.grid()
plt.show()

# 4. O‘rtacha narxlarni chiqarish
mean_prices = df.groupby('transmission')['price'].mean()
print("\nO‘rtacha narxlar:")
print(mean_prices)
```

4. Yoqilg'i turi bo'yicha avtomobillar soni:
Har bir yoqilg'i turi uchun avtomobillar sonini hisoblang va bu ma'lumotlarni bar diagrammasi orqali tasvirlang. Hozirgi kunda qaysi yoqilg'i turlari eng ommabop ekanini aniqlang.

```python
# Kodni shu yerda yozing
import pandas as pd
import matplotlib.pyplot as plt

# 1. Datasetni yuklash
df = pd.read_csv("cars_data_cleaned.csv")

# 2. fuel_type ni standartlash (kichik harf)
df['fuel_type'] = df['fuel_type'].str.lower()

# 3. Har bir yoqilg‘i turi bo‘yicha avtomobillar soni
fuel_counts = df['fuel_type'].value_counts()

print("Yoqilg'i turlari bo'yicha avtomobillar soni:")
print(fuel_counts)

# 4. Bar diagramma
plt.figure()
fuel_counts.plot(kind='bar')

plt.title("Yoqilg'i turi bo'yicha avtomobillar soni")
plt.xlabel("Yoqilg'i turi")
plt.ylabel("Avtomobillar soni")

plt.xticks(rotation=30)
plt.grid(axis='y')

plt.show()

# 5. Eng ommabop yoqilg'i turi
most_popular = fuel_counts.idxmax()
most_count = fuel_counts.max()

print(f"\nEng ommabop yoqilg'i turi: {most_popular}")
print(f"Avtomobillar soni: {most_count}")
```

 5. Dvigatel hajmi va avtomobil yoshi:
Avtomobillarning dvigatel hajmi va ishlab chiqarilgan yili o'rtasidagi bog'liqlikni tahlil qiling. Scatter plot diagrammasini tuzib, avtomobil yoshi va dvigatel hajmini ko'rsating.

```python
# Kodni shu yerda yozing
import pandas as pd
import matplotlib.pyplot as plt

# 1. Dataset yuklash
df = pd.read_csv("cars_data_cleaned.csv")

# 2. Yilni tozalash (xavfsizlik uchun)
df = df.dropna(subset=['year', 'engine_power'])

# 3. Avtomobil yoshini hisoblash
current_year = 2026
df['car_age'] = current_year - df['year']

# 4. Scatter plot
plt.figure()

scatter = plt.scatter(
    df['car_age'],        # X o‘q: avtomobil yoshi
    df['engine_power'],   # Y o‘q: dvigatel quvvati
    c=df['engine_power'], # rang: engine_power
    cmap='plasma',
    alpha=0.7
)

plt.colorbar(scatter, label="Engine Power")

plt.title("Avtomobil yoshi va dvigatel quvvati o‘rtasidagi bog‘liqlik")
plt.xlabel("Avtomobil yoshi (yil)")
plt.ylabel("Dvigatel quvvati")

plt.grid()
plt.show()

# 5. Korrelyatsiya
corr = df[['car_age', 'engine_power']].corr().iloc[0,1]
print(f"\nKorrelyatsiya: {corr:.2f}")
```

6. Narx va yoqilg'i turi o'rtasidagi bog'liqlik:
Yoqilg'i turi va avtomobil narxi o'rtasidagi bog'liqlikni o'rganing. Har bir yoqilg'i turi uchun o'rtacha narxni hisoblang va bar diagrammasini tuzing.

```python
# Kodni shu yerda yozing
import pandas as pd
import matplotlib.pyplot as plt

# 1. Datasetni yuklash
df = pd.read_csv("cars_data_cleaned.csv")

# 2. Tozalash (xavfsizlik uchun)
df = df.dropna(subset=['fuel_type', 'price'])

# 3. Yoqilg‘i turini standartlash
df['fuel_type'] = df['fuel_type'].str.lower()

# 4. Har bir yoqilg‘i turi bo‘yicha o‘rtacha narx
avg_price = df.groupby('fuel_type')['price'].mean().sort_values(ascending=False)

print("Yoqilg'i turlari bo'yicha o'rtacha narx:")
print(avg_price)

# 5. Bar diagramma
plt.figure()

avg_price.plot(kind='bar')

plt.title("Yoqilg'i turi va avtomobil narxi o‘rtasidagi bog‘liqlik")
plt.xlabel("Yoqilg'i turi")
plt.ylabel("O‘rtacha narx")

plt.xticks(rotation=30)
plt.grid(axis='y')

plt.show()

# 6. Eng qimmat yoqilg‘i turi
most_expensive = avg_price.idxmax()
print(f"\nEng qimmat yoqilg‘i turi: {most_expensive}")
```

8. Ishlab chiqarilgan yil va dvigatel hajmi:
Avtomobillarning ishlab chiqarilgan yili va dvigatel hajmi o'rtasidagi bog'liqlikni tahlil qiling. Har bir yil uchun dvigatel hajmini ko'rsatuvchi histogrammani tuzing.

```python
# Kodni shu yerda yozing
import pandas as pd
import matplotlib.pyplot as plt

# 1. Dataset yuklash
df = pd.read_csv("cars_data_cleaned.csv")

# 2. Kerakli ustunlarni tozalash
df = df.dropna(subset=['year', 'engine_power'])

# 3. Yillarni tartiblash (faqat eng muhim yillarni olish uchun)
years = sorted(df['year'].unique())

# 4. Histogram (har bir yil uchun dvigatel taqsimoti)
plt.figure(figsize=(10,6))

for y in years[:5]:  # juda ko'p bo'lmasin deb 5 ta yil
    subset = df[df['year'] == y]
    plt.hist(subset['engine_power'], bins=15, alpha=0.5, label=str(y))

# 5. Grafik dizayn
plt.title("Ishlab chiqarilgan yil va dvigatel quvvati taqsimoti")
plt.xlabel("Dvigatel quvvati")
plt.ylabel("Avtomobillar soni")
plt.legend()

plt.grid()
plt.show()
```

9. Narx va uzatish qutisi turi bo'yicha taqsimot:
Narx kategoriyalarini (past, o'rtacha, yuqori) uzatish qutisi turi (avtomat yoki mexanika) bo'yicha tahlil qiling va o'rtacha narxlarni bar diagrammasida ko'rsating.

```python
# Kodni shu yerda yozing
import pandas as pd
import matplotlib.pyplot as plt

# 1. Datasetni yuklash
df = pd.read_csv("cars_data_cleaned.csv")

# 2. Kerakli ustunlarni tozalash
df = df.dropna(subset=['transmission', 'price'])

# 3. transmission ni standartlash
df['transmission'] = df['transmission'].str.lower()

# 4. Narx kategoriyalarini yaratish (past, o‘rtacha, yuqori)
df['price_category'] = pd.qcut(
    df['price'],
    q=3,
    labels=['past', "o'rtacha", 'yuqori']
)

# 5. Guruhlash (WARNING yo‘q qilingan version)
grouped = df.groupby(
    ['transmission', 'price_category'],
    observed=True
)['price'].mean().unstack()

# 6. Natijani chiqarish
print("\nO‘rtacha narxlar jadvali:")
print(grouped)

# 7. Bar diagramma
plt.figure()

grouped.plot(kind='bar')

plt.title("Narx kategoriyasi va uzatish qutisi turi bo‘yicha tahlil")
plt.xlabel("Uzatish qutisi turi")
plt.ylabel("O'rtacha narx")

plt.xticks(rotation=0)
plt.grid(axis='y')

plt.legend(title="Narx kategoriyasi")

plt.tight_layout()
plt.show()
```
